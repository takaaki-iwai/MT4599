````python
# Vowels and consonants in pre-reformation Russian
v_pre = ['А', 'а', 'Е', 'е', 'Ѣ', 'ѣ', 'Ё', 'ё', 'И', 'и', 'Й', 'й', 'ѵ', 'Ѷ', 'I', 'i', 'О', 'о', 'У', 'у', 'Ы', 'ы', 'Э', 'э', 'Ю', 'ю', 'Я', 'я']
c_pre = ['Б', 'б', 'В', 'в', 'Г', 'г', 'Д', 'д', 'Ж', 'ж', 'З', 'з', 'К', 'к', 'Л', 'л', 'М', 'м', 'Н', 'н', 'П', 'п', 'Р', 'р', 'С', 'с', 'Т', 'т', 'Ф', 'ф', 'Ѳ', 'ѳ', 'Х', 'х', 'Ц', 'ц', 'Ч', 'ч', 'Ш', 'ш', 'Щ', 'щ']
````

````python
# Vowels and consonants in post-reformation Russian
v_post = ['А', 'а', 'Е', 'е', 'Ё', 'ё', 'И', 'и', 'О', 'о', 'У', 'у', 'Ы', 'ы', 'Э', 'э', 'Ю', 'ю', 'Я', 'я']
c_post = ['Б', 'б', 'В', 'в', 'Г', 'г', 'Д', 'д', 'Ж', 'ж', 'З', 'з', 'Й', 'й', 'К', 'к', 'Л', 'л', 'М', 'м', 'Н', 'н', 'П', 'п', 'Р', 'р', 'С', 'с', 'Т', 'т', 'Ф', 'ф', 'Х', 'х', 'Ц', 'ц', 'Ч', 'ч', 'Ш', 'ш', 'Щ', 'щ']
````

# Data Cleaning Function

````python
# DATA CLEANING FUNCTION

def novel_prep(novel, alphabet_type):
  novel0 = novel.read_text(encoding = 'utf-8')

  # Remove all characters not in the Russian alphabet
  if alphabet_type == 'pre':
    v = v_pre
    c = c_pre
  if alphabet_type == 'post':
    v = v_post
    c = c_post
  novel1 = [str(novel0) for novel0 in novel0]

  # Remove all spaces, punctuation, non-Russian letters, 'ь' and 'ъ' from the novel
  letters = v + c
  novel2 = list(filter(lambda item: item in letters, novel1))
  return(novel2)
````

# Counting and Probability Functions

````python
# FUNCTION FOR A LETTER (VOWEL-CONSONANT RATIO)

def count1(novel2, alphabet_type):
  if alphabet_type == 'pre':
    v = v_pre
    c = c_pre
  if alphabet_type == 'post':
    v = v_post
    c = c_post

  # Count the number of vowels and consonants, then calculate p and q
  # Vowels
  v_filter = list(filter(lambda item: item in v, novel2))
  v_count = len(v_filter)
  p = v_count/len(novel2)

  # Consonants
  c_filter = list(filter(lambda item: item in c, novel2))
  c_count = len(c_filter)
  q = c_count/len(novel2)

  return{"v": v_count, "p": round(p, 5), "c": c_count, "q": round(q, 5)}
````

````python
# FUNCTION FOR A LETTER AND ITS PREVIOUS LETTER (ONE-STEP TRANSITIONAL PROBABILITY)

def count2(novel2, alphabet_type):
  if alphabet_type == 'pre':
    v = v_pre
    c = c_pre
  if alphabet_type == 'post':
    v = v_post
    c = c_post

  vv = 0
  vc = 0
  cv = 0
  cc = 0

  # Use for loop to check the i th and i-1 th letters
  for i in range(1, len(novel2)):
    if novel2[i] in v:
      if novel2[i-1] in v:
        vv = vv + 1
      if novel2[i-1] in c:
        cv = cv + 1
    if novel2[i] in c:
      if novel2[i-1] in v:
        vc = vc + 1
      if novel2[i-1] in c:
        cc = cc + 1

  # Calculate probabilities
  p0 = cv/(cv + cc)
  q0 = cc/(cv + cc)
  p1 = vv/(vv + vc)
  q1 = vc/(vv + vc)

  return{"cv": cv, "p0": round(p0, 5), "cc": cc, "q0": round(q0, 5), "vv": vv, "p1": round(p1, 5), "vc": vc, "q1": round(q1, 5)}
````

````python
# FUNCTION FOR A LETTER AND ITS PREVIOUS 2 LETTERS (TWO-STEP TRANSITIONAL PROBABILITY)

def count3(novel2, alphabet_type):
  if alphabet_type == 'pre':
    v = v_pre
    c = c_pre
  if alphabet_type == 'post':
    v = v_post
    c = c_post

  vvv = 0
  vvc = 0
  vcv = 0
  vcc = 0
  cvv = 0
  cvc = 0
  ccv = 0
  ccc = 0

  # Use for loop to check the i th and i-1 th letters
  for i in range(2, len(novel2)):
    if novel2[i] in v:
      if novel2[i-1] in v:
        if novel2[i-2] in v:
          vvv = vvv + 1
        if novel2[i-2] in c:
          cvv = cvv + 1
      if novel2[i-1] in c:
        if novel2[i-2] in v:
          vcv = vcv + 1
        if novel2[i-2] in c:
          ccv = ccv + 1

    if novel2[i] in c:
      if novel2[i-1] in v:
        if novel2[i-2] in v:
          vvc = vvc + 1
        if novel2[i-2] in c:
          cvc = cvc + 1
      if novel2[i-1] in c:
        if novel2[i-2] in v:
          vcc = vcc + 1
        if novel2[i-2] in c:
          ccc = ccc + 1

  # Calculate probabilities
  p00 = ccv/(ccv + ccc)
  q00 = ccc/(ccv + ccc)
  p01 = cvv/(cvv + cvc)
  q01 = cvc/(cvv + cvc)
  p10 = vcv/(vcv + vcc)
  q10 = vcc/(vcv + vcc)
  p11 = vvv/(vvv + vvc)
  q11 = vvc/(vvv + vvc)

  return{"ccv": ccv, "p00": round(p00, 5), "ccc": ccc, "q00": round(q00, 5), "cvv": cvv, "p01": round(p01, 5), "cvc": cvc, "q01": round(q01, 5), "vcv": vcv, "p10": round(p10, 5), "vcc": vcc, "q10": round(q10, 5), "vvv": vvv, "p11": round(p11, 5), "vvc": vvc, "q11": round(q11, 5)}
````

````python
# An empty list to add counts and probabilities
data = []

# Perform each count and probability calculation, then add the values gained to data
def count(novel, alphabet_type):
  # Apply novel2 function on a given text
  novel2 = novel_prep(novel, alphabet_type)

  # Create a new row with entries writer and title of the work (extract both from the .txt file name)
  row = {"Writer": Path(novel).parent.name, "Title": Path(novel).stem}

  # Apply count1 function on the text, add the values to the row
  row.update(count1(novel2, alphabet_type))

  # Apply count2 function on the text, add the values to the row
  row.update(count2(novel2, alphabet_type))

  # Apply count3 function on the text, add the values to the row
  row.update(count3(novel2, alphabet_type))

  # Add the row to data
  data.append(row)
````

````python
# Type I error
a = 0.05
# Type II error
b = 0.10

# True values of the transitional probabilities
data = analysis1.loc["Евгенiй Онѣгинъ"]
p0_true = data["p0"]
q0_true = data["q0"]
p1_true = data["p1"]
q1_true = data["q1"]

# Sequential probability ratio test
def SPRT(n):
  onegin_n = count2(onegin_pre_prep[:n], 'pre')
  vv = onegin_n["vv"]
  p1 = onegin_n["p1"]
  vc = onegin_n["vc"]
  q1 = onegin_n["q1"]
  cv = onegin_n["cv"]
  p0 = onegin_n["p0"]
  cc = onegin_n["cc"]
  q0 = onegin_n["q0"]
  upperbound = np.log((1 - b)/a)
  lowerbound = np.log(b/(1 - a))
  LR = vv*np.log(p1/p1_true) + vc*np.log(q1/q1_true) + cv*np.log(p0/p0_true) + cc*np.log(q0/q0_true)
  return {"lowerbound": lowerbound, "upperbound": upperbound, "LR": LR}
````

# Analysis 5: Comparison within writers

````python
# Empty lists to add results for each writer
test_data1 = []
test_data2 = []
test_data3 = []

# A function that creates a row with results for a writer
def test(writer_name):
  # Find all rows of the writer in df
  writer = df_post.loc[df_post["Writer"] == writer_name]

  # Comparison of p and q
  writer1 = writer[["v", "c"]]

  # Comparison of p0, q0, p1 and q1
  writer2 = writer[["cv", "cc", "vv", "vc"]]

  # Comparison of p00, q00, p01, q01, p10, q10, p11 and q11
  writer3 = writer[["ccv", "ccc", "cvv", "cvc", "vcv", "vcc", "vvv", "vvc"]]

  # Carry out chi-squared test for writer1
  writer_r1 = chi2_contingency(writer1)

  # Calculate the expectation of p and q
  E4_p = writer["p"].mean()
  E4_q = writer["q"].mean()

  # Carry out chi-squared test for writer2
  writer_r2 = chi2_contingency(writer2)

  # Calculate the expectation of p0, q0, p1 and q1
  E4_p0 = writer["p0"].mean()
  E4_q0 = writer["q0"].mean()
  E4_p1 = writer["p1"].mean()
  E4_q1 = writer["q1"].mean()

  # Carry out chi-squared test for writer3
  writer_r3 = chi2_contingency(writer3)

  # Calculate the expectation of p00, q00, p01, q01, p10, q10, p11 and q11
  E4_p00 = writer["p00"].mean()
  E4_q00 = writer["q00"].mean()
  E4_p01 = writer["p01"].mean()
  E4_q01 = writer["q01"].mean()
  E4_p10 = writer["p10"].mean()
  E4_q10 = writer["q10"].mean()
  E4_p11 = writer["p11"].mean()
  E4_q11 = writer["q11"].mean()

  # Create a row with the writer's name
  row1 = {"Writer": writer_name}
  row2 = {"Writer": writer_name}
  row3 = {"Writer": writer_name}

  # Add the test results and expectations of p and q to the row
  row1["E(p)"] = E4_p
  row1["E(q)"] = E4_q
  row1["p-value1"] = writer_r1.pvalue
  row1["dof1"] = writer_r1.dof

  # Add the test results and expectations of p0, q0, p1 and q1 to the row
  row2["E(p0)"] = E4_p0
  row2["E(q0)"] = E4_q0
  row2["E(p1)"] = E4_p1
  row2["E(q1)"] = E4_q1
  row2["p-value2"] = writer_r2.pvalue
  row2["dof2"] = writer_r2.dof

  # Add the test results and expectations of p00, q00, p01, q01, p10, q10, p11 and q11 to the row
  row3["E(p00)"] = E4_p00
  row3["E(q00)"] = E4_q00
  row3["E(p01)"] = E4_p01
  row3["E(q01)"] = E4_q01
  row3["E(p10)"] = E4_p10
  row3["E(q10)"] = E4_q10
  row3["E(p11)"] = E4_p11
  row3["E(q11)"] = E4_q11
  row3["p-value3"] = writer_r3.pvalue
  row3["dof3"] = writer_r3.dof

  # Add the row to the test_data
  test_data1.append(row1)
  test_data2.append(row2)
  test_data3.append(row3)
````

# Analysis 6: Comparison between writers

````python
# All writers with p-value > 0.05 (critical value) for two-step transition probability
consistent_writers = results5c[results5c["p-value3"] > 0.05]

# Further filtering the writers with remove_writers list
consistent_writers = consistent_writers["Writer"].tolist()
````

````python
# Extracting rows for the filtered writers from df_post
df_consistent = df_post[df_post["Writer"].isin(consistent_writers)]

# We only need counts of vowel-consonant patterns
df_consistent = df_consistent[["Writer", "v", "c", "cv", "cc", "vv", "vc", "ccv", "ccc", "cvv", "cvc", "vcv", "vcc", "vvv", "vvc"]]

# Calculate sum of each vowel-consonant pattern for each writer
df4 = df_consistent.groupby("Writer").sum()
````

````python
def analysis6a(writer1, writer2):
  # Extract 'v' and 'c' counts for writer1 and writer2
  writer1_data = df4.loc[writer1][["v", "c"]]
  writer2_data = df4.loc[writer2][["v", "c"]]

  # Create a contingency table
  contingency_table = np.array([
      writer1_data.values,
      writer2_data.values
  ])

  # Perform chi-squared test
  return chi2_contingency(contingency_table)[1]
````

````python
# Create an empty matrix
results6a = np.zeros((len(consistent_writers), len(consistent_writers)))

# Add p-value from the analysis6a chi-squared tests of i-th and j-th writer on the consistent_writers list into the matrix entry i,j
for i in range(len(consistent_writers)):
  writer1 = consistent_writers[i]
  for j in range(len(consistent_writers)):
    writer2 = consistent_writers[j]
    results6a[i, j] = analysis6a(writer1, writer2)

# Convert into dataframe and rename the column and the index
results6a = pd.DataFrame(results6a, columns = consistent_writers)
row_labels = [consistent_writers]
results6a.index = results6a.columns
````
